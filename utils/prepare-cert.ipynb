AWS RDS requires a specific trust certificate to establish secure connections. You can either 1) disable the secure connection (not recommended) or 2) load the trust certificate. In an ephemeral workspace where it may be difficult to configure persistent certificates via standard tools/locations, you can portably bundle your existing certs with the AWS RDS cert and set an environment variable (`SSL_CERT_FILE`) to use the bundle. This notebook prepares that bundle.

**Download** the correct AWS RDS certificate bundle from here:
https://docs.aws.amazon.com/AmazonRDS/latest/UserGuide/UsingWithRDS.SSL.html#UsingWithRDS.SSL.CertificatesDownload

```python
from pathlib import Path

import certifi

with open("us-east-1-bundle.pem", "r") as f:
    rds = f.read()
base = Path(certifi.where()).read_text()
Path("ca-bundle.pem").write_text(base + "\n" + rds)
```
